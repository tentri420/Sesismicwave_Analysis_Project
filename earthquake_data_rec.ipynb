## Endpoint URL:

```python
url = "https://earthquake.usgs.gov/fdsnws/event/1/query"
```

## Record storeage:

```python
all_records=[]
start_year=datetime.now().year -5
end_year=datetime.now().year
print(end_year)
```

## Get year & month & Data record:

```python
all_records = []

for year in range(start_year, end_year + 1):

    for month in range(1, 13):

        start_date = f"{year}-{month:02d}-01"

        if month == 12:
            end_date = f"{year + 1}-01-01"
        else:
            end_date = f"{year}-{month + 1:02d}-01"

        print("Extracting:", start_date, "→", end_date)

        params = {
            "format": "geojson",
            "starttime": start_date,
            "endtime": end_date,
            "minmagnitude": 3
        }

        # Call USGS API
        response = requests.get(url, params=params)

        # Check whether request was successful
        if response.status_code == 200:

            data = response.json()

            # Extract earthquake features
            features = data["features"]

            print("Earthquakes found:", len(features))

            # Convert each earthquake into dictionary
            for feature in features:

                properties = feature["properties"]
                coordinates = feature["geometry"]["coordinates"]

                record = {
                    "id": feature["id"],
                    "time": properties["time"],
                    "updated": properties["updated"],
                    "latitude": coordinates[1],
                    "longitude": coordinates[0],
                    "depth_km": coordinates[2],
                    "mag": properties["mag"],
                    "magType": properties["magType"],
                    "place": properties["place"],
                    "status": properties["status"],
                    "tsunami": properties["tsunami"],
                    "sig": properties["sig"],
                    "net": properties["net"],
                    "nst": properties["nst"],
                    "dmin": properties["dmin"],
                    "rms": properties["rms"],
                    "gap": properties["gap"],
                    "magError": properties.get("magError"),
                    "depthError": properties.get("depthError"),
                    "magNst": properties.get("magNst"),
                    "types": properties.get("types"),
                    "ids": properties["ids"],
                    "sources": properties["sources"],
                    "type": properties["type"]
                }

                # Add dictionary to list
                all_records.append(record)

        else:
            print("API Error:", response.status_code)

# Convert list of dictionaries to DataFrame
df = pd.DataFrame(all_records)

print("\nTotal records:", len(df))
print(df.head())
print(df.shape)
```

# Cleaning datetime Fields:

```python
df.columns
```

```python
# 1. TIME FIELD:
# Already time field is cleared but not updated time:

df["time"]=pd.to_datetime(df["time"],errors="coerce")  # to convert ms to datetime format
df["updated"] = pd.to_datetime(df["updated"], errors="coerce")

#df["updated"] = df["updated"].dt.strftime("%Y-%m-%d %H:%M:%S") # to convert the datetime to string format
```

```python
df[['time','updated']].dtypes
```

```python
# To derive Year, Month , Day & Day of week from dataset:
df["time"] = pd.to_datetime(df["time"], errors="coerce")

df["year"] = df["time"].dt.year
df["month"]=df["time"].dt.month
df["day_of_week"]=df["time"].dt.day_of_week
```

```python
df["event_date"]=df["time"].dt.date
df["event_time"]=df["time"].dt.time
```

```python
#2. TEXT FIELD:
# Clear the text field using regular expression
#country = re.findall(r'[^,]+$', place)[0].strip()
import re
df["Country"] = df["place"].str.extract(r'([^,]+)$').iloc[:, 0].str.strip()
```

```python
# to normalize the column:

df["sources"] = df["sources"].str.replace(r"^,|,$", "", regex=True)
df["types"] = df["types"].str.replace(r"^,|,$", "", regex=True)
df["ids"] = df["ids"].str.replace(r"^,|,$", "", regex=True)

# To drop the column "ids" as "ID" and "IDS" are same:
df=df.drop(["ids"], axis=1)
df["id"]=df["id"].str.upper()
```

```python
#  EXTRACT DISTANCE_KM FROM PLACE:

df["distance_km"]=df["place"].str.extract(r"^(\d+)")[0].astype("Int64")
```

```python
# Fill null values with "0" in distance_km:
df["distance_km"]=df["distance_km"].fillna(0)
```

```python
df.value_counts(["status","magType","type","net","sources","types"])
```

```python
# 3.Derived Columns for depth_km & mag:

# To find shallow or depth earthquake using DEPTH_KM:


cond=[
       (df.depth_km <=70.00),
       (df.depth_km>70.00)&(df.depth_km<=300.00),
       (df.depth_km>300.00)&(df.depth_km<=700.00)
    ]
df["depth_cat"]=np.select(cond,["Shallow","Intermediate","Deep"],
       default="unknown")
```

```python
# To find the strength of earthquake using magnitude: (3 to 8)

condition=[
       (df.mag <=3.9),
       (df.mag>=4.0)&(df.mag<=4.9),
       (df.mag>=5.0)&(df.mag<=5.9),
       (df.mag>=6.0)&(df.mag<=6.9),
       (df.mag>=7.0)&(df.mag<=7.9),
       (df.mag>=8.0)
    ]
df["mag_cat"]=np.select(condition,["Minor","Light","Moderate","Strong","Major","Great"],
       default="unknown")
```

# Continent Country Conversion:

```python
#SAMPLE EXAMPLE
#print(pc.country_name_to_country_alpha2("India"))
```

```python
def get_continent(country):
    try:
        country_code = pc.country_name_to_country_alpha2(country)
        continent_code = pc.country_alpha2_to_continent_code(country_code)
        return pc.convert_continent_code_to_continent_name(continent_code)
    except:
        return "Unknown"

df["Continent"] = df["Country"].apply(get_continent)
```
